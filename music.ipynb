```python
from ffmpeg import _ffmpeg, _run
import librosa
import numpy as np
from io import BytesIO
import cv2 as cv
```

```python
def audio_from_video(video_path):
    out, _ = _run.run(
        _ffmpeg
        .input(video_path)
        .output('pipe:',format='wav'),
        capture_stdout = True,
        capture_stderr = True
    )
    
    # Lo captura desde el buffer
    return librosa.load(BytesIO(out),mono=True)
    
audio, sr = audio_from_video(r'audiomancy\common\stems\brass\4 Octave Altissimo - Baritone Saxophone (Paraschos Wooden Neck).mp4')
```
